### **3.1 ¿Qué es una Lista?**

Una lista es una **colección de elementos en un orden específico**. Puedes crear una lista con las letras del alfabeto, los dígitos del 0 al 9 o los nombres de todos los miembros de tu familia. Puedes incluir cualquier cosa en una lista, y los elementos no tienen por qué estar relacionados de ninguna manera en particular. Dado que una lista suele contener más de un elemento, es recomendable usar el plural para el nombre de la lista, como letras, dígitos o nombres.

En Python, los corchetes ([]) indican una lista, y los elementos individuales se separan con comas. Aquí tienes un ejemplo sencillo de una lista que contiene varios tipos de bicicletas:

```python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles)
```

Si le pides a Python que imprima una lista, Python devuelve su representación de la lista, incluyendo los corchetes.

Dado que este no es el resultado que quieres que vean tus usuarios, aprendamos cómo acceder a los elementos individuales de una lista.

#### **3.1.1 Acceso a Elementos en una Lista**

Las listas son **colecciones ordenadas**, por lo que puedes acceder a cualquier elemento de una lista indicándole a Python la **posición**, o **índice**, del elemento deseado. Para acceder a un elemento de una lista, escribe el nombre de la lista seguido del índice del elemento entre corchetes.

Por ejemplo, extraigamos la primera bicicleta de la lista `bicycles`:

```python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles[0])
```

Cuando solicitamos un solo elemento de una lista, Python devuelve solo ese elemento sin corchetes. Este es el resultado que quieres que vean tus usuarios: una salida limpia y con un formato impecable.

También puedes usar los métodos de cadena del Capítulo 2 en cualquier elemento de esta lista. Por ejemplo, puedes dar formato al elemento `'trek'` para que tenga una apariencia más presentable usando el método `title()`:

```python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles[0].title())
```

#### **3.1.2 Las Posiciones del Índice Comienzan en 0, no en 1**

Python considera que el **primer elemento** de una lista está en la **posición 0**, no en la 1. Esto es común en la mayoría de los lenguajes de programación, y la razón radica en cómo se implementan las operaciones de lista a un nivel inferior. Si obtienes resultados inesperados, pregúntate si estás cometiendo un error simple pero común de índice de uno.

El segundo elemento de una lista tiene un índice de 1. Usando este sistema de numeración, puedes obtener cualquier elemento de una lista restando uno a su posición en la lista. Por ejemplo, para acceder al cuarto elemento de una lista, solicitas el elemento en el índice 3.

El siguiente código solicita las bicicletas en los índices 1 y 3:

```python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles[1])
print(bicycles[3])
```

Python tiene una sintaxis especial para acceder al último elemento de una lista. Si solicitas el elemento en el **índice -1**, Python siempre devuelve el **último elemento** de la lista:

```python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles[-1])
```

Este código devuelve el valor `'specialized'`. Esta sintaxis es muy útil, ya que a menudo querrás acceder a los últimos elementos de una lista sin saber con exactitud su longitud. Esta convención también se aplica a otros índices negativos. El **índice -2** devuelve el **segundo elemento desde el final** de la lista, el **índice -3** devuelve el **tercero desde el final**, y así sucesivamente.

#### **3.1.3 Utilizando Valores Individuales de una Lista**

Puedes usar valores individuales de una lista como si fueran cualquier otra variable. Por ejemplo, puedes usar cadenas f para crear un mensaje basado en un valor de la lista.

Intentemos extraer la primera bicicleta de la lista y componer un mensaje usando ese valor:

```python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
message = f'My first bicycle was a {bicycles[0].title()}.'

print(message)
```

Construimos una oración usando el valor en `bicycles[0]` y lo asignamos a la variable `message`. La salida es una oración simple sobre la primera bicicleta de la lista.

### **3.2 Modificar, Agregar y Eliminar Elementos**

La mayoría de las listas que crees serán dinámicas, lo que significa que las irás creando y añadiendo o eliminando elementos a medida que se ejecuta el programa. Por ejemplo, podrías crear un juego en el que un jugador tenga que derribar alienígenas del cielo. Podrías almacenar el conjunto inicial de alienígenas en una lista y eliminar uno cada vez que uno sea derribado. Cada vez que aparezca un nuevo alienígena en la pantalla, lo añadirás a la lista. Tu lista de alienígenas aumentará y disminuirá de tamaño a lo largo del juego.

#### **3.2.1 Modificando Elementos en una Lista**

La sintaxis para modificar un elemento es similar a la que se usa para acceder a un elemento de una lista. Para cambiar un elemento, se utiliza el nombre de la lista seguido del índice del elemento que se desea modificar y, a continuación, se proporciona el nuevo valor que se le asignará.

Por ejemplo, supongamos que tenemos una lista de motocicletas y el primer elemento es `'honda'`. Podemos cambiar el valor de este primer elemento después de haber creado la lista:

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

motorcycles[0] = 'ducati'
print(motorcycles)
```

Aquí definimos la lista de `motocycles`, con `'honda'` como primer elemento.

Luego cambiamos el valor del primer elemento a `'ducati'`. El resultado muestra que el primer elemento ha cambiado, mientras que el resto de la lista permanece igual. Puedes cambiar el valor de cualquier elemento de una lista, no solo del primero.

#### **3.2.2 Agregar Elementos a una Lista**

Es posible que desees agregar un nuevo elemento a una lista por diversas razones. Por ejemplo, podrías querer agregar nuevos alienígenas a un juego, añadir nuevos datos a una visualización o agregar nuevos usuarios registrados a un sitio web que haya creado. Python ofrece varias maneras de agregar nuevos datos a listas existentes.

**Agregar Elementos al Final de una Lista**

La forma más sencilla de añadir un nuevo elemento a una lista es agregarlo al final (`append`) de la misma. Al agregar un elemento a una lista, este se añade al final. Usando la misma lista del ejemplo anterior, añadiremos el nuevo elemento `'ducati'` al final de la lista:

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

motorcycles.append('ducati')
print(motorcycles)
```

Aquí, el método `append()` agrega `'ducati'` al final de la lista, sin afectar a ninguno de los otros elementos de la lista.

El método `append()` facilita la creación dinámica de listas. Por ejemplo, puedes comenzar con una lista vacía y luego agregarle elementos mediante una serie de llamadas a `append()`. Partiendo de una lista vacía, agreguemos los elementos `'honda'`, `'yamaha'`y `'suzuki'`:

```python
motorcycles = []

motorcycles.append('honda')
motorcycles.append('yamaha')
motorcycles.append('suzuki')

print(motorcycles)
```

La lista resultante es exactamente igual que las listas de los ejemplos anteriores.

Crear listas de esta manera es muy común, ya que a menudo no se conocen los datos que los usuarios desean almacenar en un programa hasta que este se está ejecutando. Para que los usuarios tengan el control, comience definiendo una lista vacía que contendrá sus valores. Luego, agregue cada nuevo valor a la lista que acaba de crear.

**Insertando Elementos en una Lista**

Puedes agregar un nuevo elemento en cualquier posición de tu lista usando el método `insert()`. Para ello, debes especificar el índice del nuevo elemento y su valor:

```python
motorcycles = ['honda', 'yamaha', 'suzuki']

motorcycles.insert(0, 'ducati')
print(motorcycles)
```

En este ejemplo, insertamos el valor `'ducati'` al principio de la lista. El método `insert()` abre un espacio en la posición 0 y almacena el valor `'ducati'` en esa ubicación. Esto desplaza todos los demás valores de la lista una posición a la derecha.

#### **3.2.3 Eliminar Elementos de una Lista**

A menudo, querrás eliminar un elemento o un conjunto de elementos de una lista. Por ejemplo, cuando un jugador derriba a un alienígena, probablemente querrás eliminarlo de la lista de alienígenas activos. O cuando un usuario decide cancelar su cuenta en una aplicación web que creaste, querrás eliminarlo de la lista de usuarios activos. Puedes eliminar un elemento según su posición en la lista o según su valor.

**Eliminar un Elemento Usando la Instrucción `del`**

Si conoce la posición del elemento que desea eliminar de una lista, puede usar la instrucción `del`:

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

del motorcycles[0]
print(motorcycles)
```

Aquí utilizamos la instrucción `del` para eliminar el primer elemento, `'honda'`, de la lista `motorcycles`.

Puedes eliminar un elemento de cualquier posición en una lista usando la instrucción `del` si conoces su índice. Por ejemplo, así es como se elimina el segundo elemento, `'yamaha'`, de la lista:

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

del motorcycles[1]
print(motorcycles)
```

En ambos ejemplos, ya no se puede acceder al valor que se eliminó de la lista después de usar la instrucción `del`.

**Eliminar un Elemento con el Método pop()**

A veces, querrás usar el valor de un elemento después de eliminarlo de una lista. Por ejemplo, podrías querer obtener la posición *x* e *y* de un alienígena que acaba de ser derribado para dibujar una explosión en esa posición. En una aplicación web, podrías querer eliminar un usuario de una lista de miembros activos y luego agregarlo a una lista de miembros inactivos.

El método `pop()` elimina el último elemento de una lista, pero te permite trabajar con él después de eliminarlo. El término *pop* proviene de imaginar una lista como una pila de elementos y extraer uno de la parte superior de la pila. En esta analogía, la parte superior de una pila corresponde al final de una lista.

Eliminemos una motocicleta de la lista de motorcycles:

```python
motorcycles = ['honda', 'yamaha', 'suzuki'] #(1)
print(motorcycles)

popped_motorcycle = motorcycles.pop() #(2)
print(motorcycles) #(3)
print(popped_motorcycle) #(4)
```

Comenzamos definiendo e imprimiendo la lista de motocicletas (1). Luego, extraemos un valor de la lista y lo asignamos a la variable `poped_motorcycle` (2). Imprimimos la lista (3) para mostrar que se ha eliminado un valor. Finalmente, imprimimos el valor extraído (4) para demostrar que aún tenemos acceso al valor que se eliminó.

El resultado muestra que el valor `'suzuki'` se eliminó del final de la lista y ahora se asigna a la variable `popped_motorcycle`.

¿Cómo podría ser útil este método `pop()`? Imaginemos que las motocicletas de la lista están ordenadas cronológicamente, según la fecha en que las adquirimos. En ese caso, podemos usar el método `pop()` para imprimir información sobre la última motocicleta que compramos:

```python
motorcycles = ['honda', 'yamaha', 'suzuki']

last_owned = motorcycles.pop()
print(f'The last motorcycle I owned was a {last_owned.title()}.')
```

El resultado es una frase sencilla sobre la última motocicleta que tuvimos.

**Eliminar Elementos de Cualquier Posición en una Lista**

Puedes usar `pop()` para eliminar un elemento de cualquier posición en una lista incluyendo el índice del elemento que deseas eliminar entre paréntesis:

```python
motorcycles = ['honda', 'yamaha', 'suzuki']

first_owned = motorcycles.pop(0)
print(f'The first motorcycle I owned was a {first_owned.title()}.')
```

Comenzamos seleccionando la primera motocicleta de la lista y luego imprimimos un mensaje sobre ella. El resultado es una frase sencilla que describe la primera motocicleta que tuvimos.

Recuerda que cada vez que uses `pop()`, el elemento con el que trabajas ya no se almacena en la lista. Si no estás seguro de si usar la instrucción `del` o el método `pop()`, aquí tienes una forma sencilla de decidir: cuando quieras eliminar un elemento de una lista y no usarlo de ninguna manera, usa la instrucción `del`; si quieres usar un elemento al eliminarlo, usa el método `pop()`.

**Eliminar un Elemento por su Valor**

A veces, desconocerás la posición del elemento que deseas eliminar de una lista. Si solo conoces su valor, puedes usar el método `remove()`.

Por ejemplo, supongamos que queremos eliminar el valor `'ducati'` de la lista de motocicletas:

```python
motorcycles = ['honda', 'yamaha', 'suzuki', 'ducati']
print(motorcycles)

motorcycles.remove('ducati')
print(motorcycles)
```

Aquí, el método `remove()` le indica a Python que averigüe dónde aparece `'ducati'` en la lista y elimine ese elemento.

También puedes usar el método `remove()` para trabajar con un valor que se está eliminando de una lista. Eliminemos el valor `'ducati'` e imprimamos el motivo de su eliminación:

```python
motorcycles = ['honda', 'yamaha', 'suzuki', 'ducati'] #(1)
print(motorcycles)

too_expensive = 'ducati' #(2)
motorcycles.remove(too_expensive) #(3)
print(motorcycles)
print(f'\nA {too_expensive.title()} is too expensive for me.') #(4)
```

Después de definir la lista (1), asignamos el valor `'ducati'` a una variable llamada `too_expensive` (2). Luego usamos esta variable para indicarle a Python qué valor eliminar de la lista (3). El valor `'ducati'` se ha eliminado de la lista (4), pero aún es accesible a través de la variable `too_expensive`, lo que nos permite imprimir una declaración sobre por qué eliminamos `'ducati'` de la lista de motocicletas.

**Nota:** El método `remove()` elimina únicamente la primera aparición del valor especificado. Si existe la posibilidad de que el valor aparezca más de una vez en la lista, deberá usar un bucle para asegurarse de que se eliminen todas las apariciones.

### **3.3 Organizando una Lista**

A menudo, tus listas se crearán en un orden impredecible, ya que no siempre puedes controlar el orden en que tus usuarios proporcionan sus datos. Si bien esto es inevitable en la mayoría de los casos, con frecuencia querrás presentar tu información en un orden específico. A veces querrás conservar el orden original de tu lista, y otras veces querrás cambiarlo. Python ofrece diversas maneras de organizar tus listas, según la situación.

#### **3.3.1 Ordenar una Lista de Forma Permanente con el Método `sort()`**

El método `sort()` de Python facilita bastante la ordenación de una lista. Imaginemos que tenemos una lista de coches y queremos cambiar su orden para almacenarlos alfabéticamente. Para simplificar la tarea, supongamos que todos los valores de la lista están en minúsculas:

```python
cars = ['bmw', 'audi', 'toyota', 'subary']
cars.sort()
print(cars)
```

El método `sort()` cambia el orden de la lista de forma permanente. Los carros ahora están en orden alfabético y nunca podremos volver al orden original.

También puedes ordenar esta lista en orden alfabético inverso pasando el argumento `reverse=True` al método `sort()`. El siguiente ejemplo ordena la lista de carros en orden alfabético inverso:

```python
cars = ['bmw', 'audi', 'toyota', 'subary']
cars.sort(reverse=True)
print(cars)
```

Una vez más, el orden de la lista cambia de forma permanente.

#### **3.3.2 Ordenar una lista Temporalmente con la Función `sorted()`**

Para mantener el orden original de una lista pero presentarla ordenada, puedes usar la función `sorted()`. Esta función te permite mostrar la lista en un orden específico, pero no altera su orden original.

Probemos esta función con la lista de coches.

```python
cars = ['bmw', 'audi', 'toyota', 'subary']

print('Here is the original list:') #(1)
print(cars)

print('\nHere is the sorted list:') #(2)
print(sorted(cars))

print('\nHere is the original list again:') #(3)
print(cars)
```

Primero imprimimos la lista en su orden original (1) y luego en orden alfabético (2). Después de que la lista se muestra en el nuevo orden, mostramos que la lista todavía está almacenada en su orden original (3).

Observa que la lista permanece en su orden original después de usar la función `sorted()`. Esta función también acepta el argumento `reverse=True` si deseas mostrar la lista en orden alfabético inverso.

**Nota:**
Ordenar una lista alfabéticamente es un poco más complicado cuando no todos los valores están en minúsculas. Existen varias maneras de interpretar las mayúsculas al determinar el orden de clasificación, y especificar el orden exacto puede ser más complejo de lo que queremos abordar ahora. Sin embargo, la mayoría de los métodos de clasificación se basan directamente en lo aprendido en esta sección.

#### **3.3.3 Imprimir una Lista en Orden Inverso**

Para invertir el orden original de una lista, puedes usar el método `reverse()`. Si originalmente almacenamos la lista de autos en orden cronológico según la fecha en que los adquirimos, podríamos reorganizarla fácilmente en orden cronológico inverso:

```python
cars = ['bmw', 'audi', 'toyota', 'subary']
print(cars)

cars.reverse()
print(cars)
```

Observa que `reverse()` no ordena alfabéticamente hacia atrás; simplemente invierte el orden de la lista.

El método `reverse()` cambia el orden de una lista de forma permanente, pero se puede volver al orden original en cualquier momento aplicando `reverse()` a la misma lista por segunda vez.

#### **3.3.4 Determinando la Longitud de una Lista**

Puedes averiguar rápidamente la longitud de una lista usando la función `len()`. La lista de este ejemplo tiene cuatro elementos, por lo que su longitud es 4:

```python
cars = ['bmw', 'audi', 'toyota', 'subary']
len(cars)
```

La función `len()` te resultará útil cuando necesites identificar el número de alienígenas que aún quedan por derribar en un juego, determinar la cantidad de datos que debes gestionar en una visualización o averiguar el número de usuarios registrados en un sitio web, entre otras tareas.

### **3.4 Evitando Errores de Índice al Trabajar con Listas**

Hay un tipo de error común que se suele ver cuando se trabaja con listas por primera vez. Digamos que tienes una lista con tres elementos y pides el cuarto:

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles[3])
```

Este ejemplo produce un *error de índice* (index error).

Python intenta devolverte el elemento en el índice 3. Pero al buscar en la lista, ningún elemento de la lista `motorcycles` tiene el índice 3. Debido a la naturaleza de indexación de listas con un desfase de uno, este error es común. La gente suele pensar que el tercer elemento es el número 3, porque empiezan a contar desde 1. Pero en Python, el tercer elemento es el número 2, porque empieza a indexar desde 0.

Un error de índice significa que Python no puede encontrar un elemento en el índice solicitado. Si se produce un error de índice en tu programa, intenta ajustar el índice solicitado en uno. Luego, ejecuta el programa nuevamente para comprobar si los resultados son correctos.

Recuerda que siempre que quieras acceder al último elemento de una lista, debes usar el índice -1. Esto siempre funcionará, incluso si el tamaño de la lista ha cambiado desde la última vez que accediste a ella:

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles[-1])
```

El índice -1 siempre devuelve el último elemento de una lista, en este caso el valor `'suzuki'`.

Este método solo provocará un error si se solicita el último elemento de una lista vacía:

```python
motorcycles = []
print(motorcycles[-1])
```

No hay artículos en motocicletas, por lo que Python devuelve otro error de índice.

Si se produce un error de índice y no logras resolverlo, intenta imprimir la lista o simplemente su longitud. Es posible que la lista se vea muy diferente de lo que esperabas, especialmente si tu programa la ha gestionado dinámicamente. Ver la lista real, o el número exacto de elementos, puede ayudarte a solucionar este tipo de errores lógicos.
