#### **2.1.1 Nombrando y Usando Variables**

Al usar variables en Python, es necesario seguir algunas reglas y pautas para evitar errores; otras pautas simplemente te ayudarán a escribir código más fácil de leer y comprender. Asegúrate de tener en cuenta las siguientes reglas al trabajar con variables:

- Los nombres de las variables solo pueden contener letras, números y guiones bajos. Pueden comenzar con una letra o un guion bajo, pero no con un número. Por ejemplo, puedes llamar a una variable `mensaje_1`, pero no `1_mensaje`.

- No se permiten espacios en los nombres de las variables, pero se pueden usar guiones bajos para separar palabras. Por ejemplo, `saludo_mensaje` funciona, pero `saludo mensaje` provocará errores.

- Evita usar *palabras clave* y nombres de funciones de Python como nombres de variables. Por ejemplo, no uses la palabra `print` como nombre de variable; Python la ha reservado para un propósito programático específico.

- Los nombres de las variables deben ser cortos pero descriptivos. Por ejemplo, `nombre` es mejor que `n`, `nombre_estudiante` es mejor que `s_n`, y `longitud_nombre` es mejor que `longitud_del_nombre_de_la_persona`.

- Tenga cuidado al usar la letra *l* minúscula y la letra *O* mayúscula, ya que podrían confundirse con los números *1* y *0*.

### **2.2 Cadenas**

Los programas organizan la información en diferentes tipos de datos para poder procesarla y utilizarla de forma útil. El primer dato que analizaremos es la cadena de caracteres.

* **¿Qué es una cadena?** La cadena de caracteres (*string*) es una secuencia de texto que, a pesar de su aparente simplicidad, es muy versátil y tiene múltiples aplicaciones.
* **Regla en Python:** En este lenguaje, cualquier elemento que esté encerrado entre **comillas simples (`' '`) o dobles (`" "`)** se considera automáticamente una cadena de caracteres.

```python
"Esto es una Cadena"
'Esto también es una Cadena'
```

#### **2.2.1 Cambiar Mayúsculas y Minúsculas en una Cadena con Métodos**

Una de las tareas más sencillas que puedes realizar con cadenas es cambiar las mayúsculas y minúsculas de las palabras. Observa el siguiente código e intenta comprender qué sucede:

```python
nombre = "ada lovelace"
print(nombre.title())
```

En este ejemplo, la variable `name` se refiere a la cadena en minúsculas "ada lovelace". El método `title()` aparece después de la variable en la llamada a `print()`. Un método es una acción que Python puede realizar sobre un dato. El punto (.) después de `name` en `name.title()` le indica a Python que el método `title()` actúe sobre la variable `name`.

Cada método va seguido de un paréntesis, ya que a menudo necesita información adicional para funcionar. Esta información se proporciona dentro de los paréntesis. La función `title()` no necesita información adicional, por lo que sus paréntesis están vacíos.

También hay otros métodos útiles disponibles para manejar mayúsculas y minúsculas. Por ejemplo, puedes convertir una cadena a mayúsculas o minúsculas de la siguiente manera:

```python
nombre = "Ada Lovelace"
print(nombre.upper())
print(nombre.lower())
```

#### **2.2.2 Uso de Variables en Cadenas de Texto**

En algunas situaciones, querrás usar el valor de una variable dentro de una cadena. Por ejemplo, podrías usar dos variables para representar un nombre y un apellido, respectivamente, y luego combinar esos valores para mostrar el nombre completo de una persona:

```python
first_name = 'ada'
last_name = 'lovelace'
full_name = f'{first_name} {last_name}'
print(full_name)
```

Para insertar el valor de una variable en una cadena, coloca la letra f justo antes de la comilla de apertura. Encierra entre llaves el nombre o los nombres de las variables que quieras usar dentro de la cadena. Python reemplazará cada variable con su valor al mostrar la cadena.

Estas cadenas se denominan *f-strings*. La *f* significa *formato*, ya que Python formatea la cadena reemplazando el nombre de cualquier variable entre llaves por su valor.

Se pueden hacer muchas cosas con las *f-strings*. Por ejemplo, se pueden usar *f-strings* para componer mensajes completos utilizando la información asociada a una variable, como se muestra aquí:

```python
first_name = 'ada'
last_name = 'lovelace'
full_name = f'{first_name} {last_name}'
print(f'Hola, {full_name.title()}!')
```

El nombre completo se usa en una frase que saluda al usuario, y el método `title()` cambia el nombre a formato de título. Este código devuelve un saludo sencillo pero bien formateado.

También puedes usar *f-strings* para componer un mensaje y luego asignar el mensaje completo a una variable:

```python
first_name = 'ada'
last_name = 'lovelace'
full_name = f'{first_name} {last_name}'
message = f'Hola, {full_name.title()}!'
print(message)
```

Este código también muestra el mensaje **¡Hola, Ada Lovelace!**, pero al asignar el mensaje a una variable hacemos que la llamada final a `print()` sea mucho más sencilla.

#### **2.2.3 Agregar Espacios en Blanco a las Cadenas con Tabulaciones o Saltos de Línea**

En programación, los espacios en blanco se refieren a cualquier carácter no imprimible, como *espacios*, *tabulaciones* y *símbolos de fin de línea*. Puedes usar espacios en blanco para organizar tu código y facilitar su lectura.

Para agregar una tabulación a su texto, utilice la combinación de caracteres `\t`:

```python
print('Python')
print('\tPython')
```

Para agregar un salto de línea en una cadena, utilice la combinación de caracteres `\n`:

```python
print('Languages:\nPython\nC\nJavaScript')
```

También puedes combinar tabulaciones y saltos de línea en una sola cadena. La cadena `"\n\t"` le indica a Python que pase a una nueva línea y que la siguiente comience con una tabulación. El siguiente ejemplo muestra cómo puedes usar una cadena de una sola línea para generar cuatro líneas de salida:

```python
print('Languages:\n\tPython\n\tC\n\tJavaScript')
```

#### **2.2.4 Eliminando Espacios en Blanco**

Los espacios en blanco adicionales pueden resultar confusos en tus programas. Para los programadores, "python" y "python ' parecen prácticamente iguales. Pero para un programa, son dos cadenas diferentes. Python detecta el espacio adicional en "python ' y lo considera significativo a menos que le indiques lo contrario.

Es importante tener en cuenta los espacios en blanco, ya que a menudo se querrá comparar dos cadenas para determinar si son iguales. Los espacios en blanco adicionales también pueden resultar confusos en situaciones mucho más sencillas. Afortunadamente, Python facilita la eliminación de espacios en blanco adicionales de los datos que introducen las personas.

Python puede buscar espacios en blanco adicionales a la derecha y a la izquierda de una cadena. Para asegurarse de que no existan espacios en blanco a la derecha de una cadena, utiliza el método `rstrip()`:

```python
favorite_language = 'Python '
favorite_language
```

El valor asociado a `favorite_language` contiene un espacio en blanco adicional al final de la cadena. Al solicitar este valor a Python en una sesión de terminal, se puede observar dicho espacio.

Cuando el método `rstrip()` actúa sobre la variable `favorite_language`, este espacio adicional se elimina.

```python
# Imprimir eliminando espacios en blanco finales con rstrip()
favorite_language.rstrip()
```

Sin embargo, solo se elimina temporalmente. Si se solicita el valor de `favorite_language` nuevamente, la cadena se verá igual que cuando se ingresó, incluyendo el espacio en blanco adicional.

```python
favorite_language
```

Para eliminar el espacio en blanco de la cadena de forma permanente, es necesario asociar el valor sin espacios en blanco con el nombre de la variable.

```python
favorite_language = 'python '
favorite_language = favorite_language.rstrip()
favorite_language
```

Para eliminar los espacios en blanco de una cadena, se eliminan de la parte derecha y se asocia este nuevo valor con la variable original. , o de ambos lados a la vez usando `strip()`.

```python
# Cadena con espacios en blanco iniciales y finales
favorite_language = ' python '
favorite_language.rstrip() # Quitar espacios en blanco finales con rstrip()
```

También se pueden eliminar los espacios en blanco de la parte izquierda de una cadena usando el método `lstrip()`

```python
favorite_language.lstrip() # Quitar espacio en blanco iniciales con lstrip()
```

O puedes eliminar los espacios en blanco de ambos lados a la vez usando `strip()`.

```python
favorite_language.strip() # Quitar espacios iniciales y finales con strip()
```

En el ejemplo anterior, partimos de un valor que contiene espacios en blanco al principio y al final. Luego, eliminamos los espacios adicionales del lado derecho, del lado izquierdo y de ambos lados. Experimentar con estas funciones de eliminación de espacios en blanco puede ayudarte a familiarizarte con la manipulación de cadenas. En la práctica, estas funciones se utilizan con mayor frecuencia para limpiar la entrada del usuario antes de almacenarla en un programa.

Primero, veamos cómo Python maneja los números enteros, ya que son los más sencillos de usar.

#### **2.3.1 Enteros**

En Python puedes sumar (+), restar (-), multiplicar (*) y dividir (/) números enteros.

```python
# Sumar enteros
2 + 3
```

```python
# Restar enteros
3 - 2
```

#### **2.3.2 Números en Coma Flotante (Floats)**

Python denomina número de coma flotante a cualquier número con decimales. Este término se utiliza en la mayoría de los lenguajes de programación y se refiere a que el punto decimal puede aparecer en cualquier posición de un número. En general, puedes usar números de punto flotante sin preocuparte por su comportamiento. Simplemente introduce los números que quieras usar y Python probablemente hará lo que esperas.

```python
0.1 + 0.1
```

```python
0.2 + 0.2
```

```python
2 * 0.1
```

```python
2 * 0.2
```

Ten en cuenta que a veces puedes obtener un número arbitrario de decimales en tu respuesta:

```python
0.2 + 0.1
```

```python
3 * 0.1
```

Esto ocurre en todos los lenguajes y no suele ser motivo de preocupación. Python intenta representar el resultado con la mayor precisión posible, lo cual a veces resulta difícil dada la forma en que los ordenadores representan los números internamente.

#### **2.3.3 Números Enteros y en Coma Flotante**

Cuando divides dos números cualesquiera, incluso si son enteros que dan como resultado un número entero, siempre obtendrás un número decimal:

```python
4/2
```

Si mezclas un número entero y un número decimal en cualquier otra operación, también obtendrás un número decimal:

```python
1 + 2.0
```

```python
2 * 3.0 
```

```python
3.0 ** 2
```

Python utiliza por defecto un número de coma flotante en cualquier operación que utilice este tipo de dato, incluso si el resultado es un número entero.

#### **2.3.4 Subrayados en los Números**

Cuando escribas números largos, puedes agrupar los dígitos usando guiones bajos para que los números grandes sean más legibles:

```python
universe_age = 14_000_000_000
```

```python
print(universe_age)
```

Python ignora los guiones bajos al almacenar este tipo de valores. Incluso si no agrupas los dígitos de tres en tres, el valor no se verá afectado. Para Python, `1000` es lo mismo que `1_000`, que a su vez es lo mismo que `10_00`. Esta característica funciona tanto para números enteros como para números de coma flotante.

### **2.4 Comentarios**

Los comentarios son una función extremadamente útil en la mayoría de los lenguajes de programación. Todo lo que has escrito en tus programas hasta ahora es código Python. A medida que tus programas se vuelven más largos y complejos, deberías añadir **notas que describan tu enfoque general para resolver el problema**. Un comentario te permite escribir notas en tu propio lenguaje, dentro de tus programas.

#### **2.4.1 Cómo se Escriben los Comentarios**

En Python, el símbolo de numeral (#) indica un comentario. Todo lo que sigue a un símbolo de numeral en tu código es ignorado por el intérprete de Python. Por ejemplo:

```python
# Decir hola a todos

print('Hello Python people!')
```

Python ignora la primera línea y ejecuta la segunda.

#### **2.4.2 ¿Qué Tipo de Comentarios Deberías Escribir?**

* La principal razón para escribir comentarios es *explicar qué se supone que hace tu código y cómo lo estás implementando*. Cuando estás trabajando en un proyecto, entiendes cómo encajan todas las piezas.

* Pero cuando retomas un proyecto después de un tiempo, es probable que hayas olvidado algunos detalles. Siempre puedes estudiar tu código durante un tiempo y averiguar cómo se suponía que funcionaban ciertas partes, pero escribir buenos comentarios puede ahorrarte tiempo al resumir claramente tu enfoque general.

* Si quieres convertirte en programador profesional o colaborar con otros programadores, debes escribir *comentarios significativos*. Hoy en día, la mayoría del software se desarrolla de forma colaborativa, ya sea por un grupo de empleados de una empresa o por un grupo de personas que trabajan juntas en un proyecto de código abierto. Los programadores experimentados esperan ver comentarios en el código, así que lo mejor es empezar a *añadir comentarios descriptivos* a tus programas desde ahora. Escribir *comentarios claros y concisos* en tu código es uno de los hábitos más beneficiosos que puedes adquirir como programador principiante.

* Cuando decidas si escribir un comentario, pregúntate si tuviste que considerar varias opciones antes de encontrar una solución razonable; si es así, escribe un comentario sobre tu solución. Es mucho más fácil eliminar comentarios adicionales después que volver atrás y escribir comentarios para un programa con pocos comentarios.

### **2.5 El Zen de Python**

Los programadores de Python con experiencia te animarán a evitar la complejidad y a buscar la simplicidad siempre que sea posible. La filosofía de la comunidad Python se encuentra en **«El Zen de Python» de Tim Peters**. Puedes acceder a este breve conjunto de principios para escribir buen código Python introduciendo `import this` en tu intérprete.

`Beautiful is better than ugly` (Hermoso es mejor que feo)

- Los programadores de Python aceptan la idea de que el código puede ser bello y elegante. En programación, se resuelven problemas. Los programadores siempre han valorado las soluciones bien diseñadas, eficientes e incluso bellas.
- A medida que aprendas más sobre Python y lo uses para escribir más código, es posible que un día alguien te mire por encima del hombro y diga: «¡Guau, qué código tan bonito!».

`Simple is better than complex` (Simple es mejor que complejo)

- Si tienes que elegir entre una solución simple y una compleja, y ambas funcionan, usa la simple. Tu código será más fácil de mantener y, tanto tú como otros podrán desarrollarlo posteriormente con mayor facilidad.


`Complex is better than complicated` (Complejo es mejor que complicado)

- La vida real es complicada, y a veces una solución sencilla a un problema resulta inalcanzable. En ese caso, utilice la solución más simple que funcione.

`Readability counts` (La legibilidad es fundamental)

* Incluso cuando tu código sea complejo, procura que sea legible. Cuando trabajes en un proyecto que implique programación compleja, concéntrate en escribir comentarios informativos para ese código.

`There should be one-- and preferably only one --obvious way to do it` (Debería haber una —y preferiblemente solo una— manera obvia de hacerlo.)

* Si se les pide a dos programadores de Python que resuelvan el mismo problema, deberían encontrar soluciones bastante compatibles. Esto no significa que no haya espacio para la creatividad en la programación. ¡Al contrario, hay mucho espacio para ella! Sin embargo, gran parte de la programación consiste en utilizar enfoques sencillos y comunes para situaciones simples dentro de un proyecto más amplio y creativo. Los detalles de tus programas deben ser comprensibles para otros programadores de Python.

`Now is better than never` (Más vale tarde que nunca)

* Podrías pasarte la vida entera aprendiendo todas las complejidades de Python y de la programación en general, pero nunca terminarías ningún proyecto. No intentes escribir código perfecto; escribe código que funcione y luego decide si mejorarlo para ese proyecto o pasar a algo nuevo.

A medida que avances y profundices en temas más complejos, ten presente esta filosofía de simplicidad y claridad. Los programadores experimentados valorarán más tu código y estarán encantados de darte su opinión y colaborar contigo en proyectos interesantes.
