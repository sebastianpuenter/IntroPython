```python
import random
import matplotlib.pyplot as plt

# 1. Definir las asignaturas
asignaturas = ['Matemáticas', 'Historia', 'Ciencias', 'Inglés', 'Arte']

# 2. Crear la lista de promedios aleatorios (entre 1.0 y 10.0)
promedios = []
for i in range(5):
    nota_aleatoria = random.uniform(1.0, 10.0)
    promedios.append(round(nota_aleatoria, 1))

# 3. Clasificar por 4 niveles y asignar colores (de Rojo a Verde)
colores_barras = []
for promedio in promedios:
    if promedio < 6.0:
        colores_barras.append('crimson')      # Bajo (Rojo)
    elif promedio < 8.0:
        colores_barras.append('gold')         # Básico (Amarillo)
    elif promedio < 9.5:
        colores_barras.append('limegreen')    # Alto (Verde claro)
    else:
        colores_barras.append('darkgreen')    # Superior (Verde oscuro)

# 4. Crear el gráfico de barras (zorder=3 para que queden al frente)
plt.bar(asignaturas, promedios, color=colores_barras, edgecolor='black', width=0.6, zorder=3)

# 5. PINTAR EL FONDO (Misma escala de colores con opacidad muy suave)
plt.axhspan(0, 6.0, facecolor='crimson', alpha=0.07, label='Bajo (0.0 - 5.9)')
plt.axhspan(6.0, 8.0, facecolor='gold', alpha=0.07, label='Básico (6.0 - 7.9)')
plt.axhspan(8.0, 9.5, facecolor='limegreen', alpha=0.07, label='Alto (8.0 - 9.4)')
plt.axhspan(9.5, 10.0, facecolor='darkgreen', alpha=0.07, label='Superior (9.5 - 10.0)')

# 6. Configurar títulos y límites del eje Y
plt.title('Rendimiento Escolar por Asignatura (Escala Rojo-Verde)')
plt.xlabel('Asignaturas')
plt.ylabel('Promedio Final')
plt.ylim(0, 10)

# 7. Mostrar la leyenda del gráfico
plt.legend(title="Niveles de Logro", loc='upper left')

# 8. Mostrar el gráfico definitivo
plt.show()
```
