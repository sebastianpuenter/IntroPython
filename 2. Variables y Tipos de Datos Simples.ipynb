### **2.1 Variables**
Vamos a usar una variable en nuestro primer programa:

```python
mensaje = 'Hola Mundo Python!'
print(mensaje)
```

Se añadió una variable llamada `mensaje`. Cada variable está asociada a un valor, que es la información que contiene. En este caso, el valor es el texto "¡Hola mundo Python!".

Ampliemos nuestro programa para imprimir un segundo mensaje:

```python
# Añadimos un segundo mensaje

mensaje = 'Hola Mundo Python!'
print(mensaje)

mensaje = 'Dios te Bendiga!'
print(mensaje)
```

Puedes cambiar el valor de una variable en tu programa en cualquier momento, y Python siempre mantendrá un registro de su valor actual.

#### **2.1.1 Nombrando y usando Variables**
Al usar variables en Python, es necesario seguir algunas reglas y pautas para evitar errores; otras pautas simplemente te ayudarán a escribir código más fácil de leer y comprender. Asegúrate de tener en cuenta las siguientes reglas al trabajar con variables:

- Los nombres de las variables solo pueden contener letras, números y guiones bajos. Pueden comenzar con una letra o un guion bajo, pero no con un número. Por ejemplo, puedes llamar a una variable `mensaje_1`, pero no `1_mensaje`.

- No se permiten espacios en los nombres de las variables, pero se pueden usar guiones bajos para separar palabras. Por ejemplo, `saludo_mensaje` funciona, pero `saludo mensaje` provocará errores.

- Evita usar *palabras clave* y nombres de funciones de Python como nombres de variables. Por ejemplo, no uses la palabra `print` como nombre de variable; Python la ha reservado para un propósito programático específico.

- Los nombres de las variables deben ser cortos pero descriptivos. Por ejemplo, `nombre` es mejor que `n`, `nombre_estudiante` es mejor que `s_n`, y `longitud_nombre` es mejor que `longitud_del_nombre_de_la_persona`.

- Tenga cuidado al usar la letra *l* minúscula y la letra *O* mayúscula, ya que podrían confundirse con los números *1* y *0*.

#### **2.1.2 Cómo Evitar Errores de Nombres al Usar Variables**

Los errores al escribir nombres de variables son muy comunes en programación, especialmente al comenzar a aprender Python. Cuando una variable se escribe incorrectamente, el programa genera un error porque Python no reconoce ese nombre. Sin embargo, el intérprete de Python ayuda al programador mostrando un “rastreo de pila” o traceback, que indica en qué parte del código ocurrió el problema. Aprender a leer estos mensajes de error es importante para identificar y corregir errores de manera rápida y eficiente.

```python
# En este programa se introduce un error a propósito en el nombre de la variable
mensaje = 'Hola, Sigue aprendiento Python!'
print(mensje)
```

La salida indica que se produjo un error en la línea 2 de la celda. El intérprete muestra esta línea 2 para ayudarnos a detectar el error rápidamente e indica qué tipo de error encontró. En este caso, se detectó un error de nombre e informa que la variable que se imprime, `mensje`, no está definida.

Python no puede identificar el nombre de variable proporcionado. Un error de nombre generalmente significa que olvidamos asignar un valor a la variable antes de usarla o que cometimos un error ortográfico al escribir su nombre. Si Python encuentra un nombre de variable similar al que no reconoce, preguntará si ese es el nombre que se pretendía usar.

En este ejemplo, omitimos la letra *s* en el nombre de la variable `mensaje` en la segunda línea. El intérprete de Python no revisa la ortografía del código, pero sí se asegura de que los nombres de las variables se escriban de forma consistente.

### **2.2 Cadenas**

Los programas organizan la información en diferentes tipos de datos para poder procesarla y utilizarla de forma útil. El primer dato que analizaremos es la cadena de caracteres.

* **¿Qué es una cadena?** La cadena de caracteres (*string*) es una secuencia de texto que, a pesar de su aparente simplicidad, es muy versátil y tiene múltiples aplicaciones.
* **Regla en Python:** En este lenguaje, cualquier elemento que esté encerrado entre **comillas simples (`' '`) o dobles (`" "`)** se considera automáticamente una cadena de caracteres.

```python
"Esto es una Cadena"
'Esto también es una Cadena'
```

#### **2.2.1 Cambiar mayúsculas y minúsculas en una cadena con métodos**

Una de las tareas más sencillas que puedes realizar con cadenas es cambiar las mayúsculas y minúsculas de las palabras. Observa el siguiente código e intenta comprender qué sucede:

```python
nombre = "ada lovelace"
print(nombre.title())
```

En este ejemplo, la variable `name` se refiere a la cadena en minúsculas "ada lovelace". El método `title()` aparece después de la variable en la llamada a `print()`. Un método es una acción que Python puede realizar sobre un dato. El punto (.) después de `name` en `name.title()` le indica a Python que el método `title()` actúe sobre la variable `name`.

Cada método va seguido de un paréntesis, ya que a menudo necesita información adicional para funcionar. Esta información se proporciona dentro de los paréntesis. La función `title()` no necesita información adicional, por lo que sus paréntesis están vacíos.

También hay otros métodos útiles disponibles para manejar mayúsculas y minúsculas. Por ejemplo, puedes convertir una cadena a mayúsculas o minúsculas de la siguiente manera:

```python
nombre = "Ada Lovelace"
print(nombre.upper())
print(nombre.lower())
```
